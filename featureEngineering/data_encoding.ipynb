```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder ## don't use the ohe when there are multiple category
```

```python
## create a simple dataframe

df=pd.DataFrame({'color':['red','blue','green','green','red','blue']})
```

```python
df.head()
```

```python
## create an instance of one hot encoder
encoder=OneHotEncoder()
```

```python
## perform fit and transform 
encoder.fit_transform(df[['color']]).toarray()
```

```python
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
## for new data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

```python
import seaborn as sns
df=sns.load_dataset('tips')
df.tail()
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoded=encoder.fit_transform(df[['sex','smoker','day']]).toarray()
```

```python
import pandas as pd
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
df_dropped=df.drop(['sex','smoker','day'],axis=1)
```

```python
final_df=pd.concat([df_dropped,encoder_df],axis=1)

print(final_df.head())
```
